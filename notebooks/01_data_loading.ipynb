```python
def check_df(df):
    print("HEAD")
    display(df.head())
    
    print("\nSHAPE")
    print(df.shape)
    
    print("\nCOLUMNS")
    print(df.columns.tolist())
    
    print("\nINFO")
    df.info()
    
    print("\nMISSING VALUES")
    print(df.isnull().sum())
    
    print("\nDUPLICATES")
    print(df.duplicated().sum())

check_df(df)
```

```python
# Investigating time-related columns
admission_dates = pd.to_datetime(df["Date of Admission"], errors="coerce")
discharge_dates = pd.to_datetime(df["Discharge Date"], errors="coerce")
```

```python
# Check for empty dates
print("Invalid admission dates:", admission_dates.isna().sum())
print("Invalid discharge dates:", discharge_dates.isna().sum())
```

```python
length_of_stay = discharge_dates - admission_dates

length_of_stay.describe()
```

```python
print("Negative stays:", (length_of_stay < pd.Timedelta(0)).sum())
print("Same-day stays:", (length_of_stay == pd.Timedelta(0)).sum())
```
