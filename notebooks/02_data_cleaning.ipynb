#### STEP 1: Load the dependencies and data

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/healthcare_dataset.csv")

print("Shape:", df.shape)
```

#### STEP 2: Data Cleaning

```python
def clean_df(df):
    df = df.copy()
    
    # Remove exact duplicate rows
    df = df.drop_duplicates()
    
    # Convert date columns to datetime
    df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
    df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
    
    # Convert negative billing amounts to missing values
    df.loc[df["Billing Amount"] < 0, "Billing Amount"] = pd.NA

     # Reset index after removing rows
    df = df.reset_index(drop=True)

    return df
```

```python
cleaned_df = clean_df(df)

print("Shape:", cleaned_df.shape)

print(
    "Negative billing amounts:",
    (cleaned_df["Billing Amount"] < 0).sum()
)

print(
    "Missing billing amounts:",
    cleaned_df["Billing Amount"].isna().sum()
)

print(
    "info:",
    cleaned_df.info()
)

print("Missing values:", cleaned_df.isna().sum())
```

```python
cleaned_df.columns
```

```python
# Standardize column names to snake_case
cleaned_df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r"[^a-zA-Z0-9]+", "_", regex=True)
    .str.strip("_")
)

# Standardize names to proper title case
cleaned_df["name"] = cleaned_df["name"].str.strip().str.title()
```

#### STEP 3: Save the clean data

```python
from pathlib import Path

Path("../data/processed").mkdir(parents=True, exist_ok=True)

cleaned_df.to_csv(
    "../data/processed/cleaned_healthcare.csv",
    index=False
)
```
